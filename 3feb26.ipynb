Develop a Python program to check whether the given string is a palindrome without using inbuilt functions or reverse indexing.

```python
print("Enter a string: ")
s = input()

is_palindrome = True
n = len(s)

for i in range(n // 2):
    if s[i] != s[n - i - 1]:
        is_palindrome = False
        break

if is_palindrome:
    print("'" + s + "' is a Palindrome.")
else:
    print("'" + s + "' is NOT a Palindrome.")
```
