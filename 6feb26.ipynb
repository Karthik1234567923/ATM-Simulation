You are analyzing user reviews.

The program should accept a paragraph of text as input and ignore all punctuation. It should find all unique words in the paragraph and print:
The count of unique words.
The unique words themselves in alphabetical order, separated by spaces.
Input Format

A single line containing the paragraph.

Constraints

NA

Output Format

First line: count of unique words.
Second line: the unique words in sorted alphabetical order, separated by spaces.
Sample Input 0

Hello world! Hello everyone.
Sample Output 0

3
Hello everyone world
Sample Input 1

Python is great. Python is fun.
Sample Output 1

4
Python fun great is

```python
import string

# Step 1: Read input
text = input()

# Step 2: Remove punctuation
for ch in string.punctuation:
    text = text.replace(ch, "")

# Step 3: Split into words
words = text.split()

# Step 4: Get unique words
unique_words = set(words)

# Step 5: Sort words
sorted_words = sorted(unique_words)

# Step 6: Output
print(len(sorted_words))
print(*sorted_words)
```
