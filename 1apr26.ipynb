13. Develop a Python program to perform string operations for the given string.

```python
s = input("Enter a string: ")

print("Length:", len(s))
print("Uppercase:", s.upper())
print("Lowercase:", s.lower())
print("Reverse:", s[::-1])
print("Count of 'a':", s.count('a'))
```
