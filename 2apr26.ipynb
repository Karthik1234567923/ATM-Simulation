Python Program – String Operations
# Logic

1. Take a string from user
2.Perform operations:

Display string
Find length
Convert to uppercase
Convert to lowercase
Reverse string
Count a character

```python
s = input("Enter a string: ")

print("String:", s)
print("Length:", len(s))
print("Uppercase:", s.upper())
print("Lowercase:", s.lower())
print("Reverse:", s[::-1])
print("Count of 'a':", s.count('a'))
```
