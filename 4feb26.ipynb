Palindrome Check (No Inbuilt Functions)

```python
print("Enter a string: ")
s = input()

flag = True
n = len(s)

for i in range(n // 2):
    if s[i] != s[n - i - 1]:
        flag = False
        break

if flag:
    print("'" + s + "' is a Palindrome.")
else:
    print("'" + s + "' is NOT a Palindrome.")
```
