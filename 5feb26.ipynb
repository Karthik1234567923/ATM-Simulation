6. Strings – String Operations

```python
print("Enter a string: ")
s = input()

print("Original:", s)
print("Length:", len(s))
print("Uppercase:", s.upper())
print("Lowercase:", s.lower())

if len(s) >= 4:
    print("Slice [1:4]:", s[1:4])
else:
    print("String too short for slicing [1:4]")

print("Count of 'a':", s.count('a'))
print("Starts with 'He':", s.startswith("He"))
```
